```python
#group the data by democracy_type and calculate the average democracy_score
```

```python
democracy_type_set = df['democracy_type'].value_counts()
print('The group of the countries based on democracy type is as:',democracy_type_set)
print('\n The average democracy score of all  the countries in the world is:',df['democracy_score'].mean().round(2))
```
